### Task 2 – Data Cleaning & Preparation 

- Handle missing values

- Validate Quantity, OrderValue, TotalAmount

```python
# Check missing values
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

```python
# Handle missing values
if df.isnull().sum().sum() == 0:
    print("No missing values found.")
else:
    df = df.dropna()
    print("Missing values were removed.")

print("Dataset shape:", df.shape)
```

```python
# Check duplicate rows
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
# Remove duplicate rows
df = df.drop_duplicates()

print("Dataset shape after removing duplicates:", df.shape)
```

```python
print("Quantity statistics:")
print(df["Quantity"].describe())

print("\nInvalid Quantity values:")
print(df[df["Quantity"] <= 0][["OrderID", "Quantity"]])
```

```python
print("OrderValue statistics:")
print(df["OrderValue"].describe())

print("\nInvalid OrderValue values:")
print(df[df["OrderValue"] <= 0][["OrderID", "OrderValue"]])
```

```python
# Calculate expected TotalAmount
expected_total = df["OrderValue"] * df["Quantity"]

# Check whether calculated value matches TotalAmount
mismatch = (df["TotalAmount"] - expected_total).abs() > 0.01

print("Number of incorrect TotalAmount values:", mismatch.sum())
```

```python
basket = (
    df.groupby(["CustomerID", "FoodItem"])["Quantity"]
    .sum()
    .unstack(fill_value=0)
)

# Convert quantities into 0 and 1
basket = (basket > 0).astype(int)

print("Basket shape:", basket.shape)
basket.head()
```

```python
frequent_itemsets = apriori(
    basket,
    min_support=0.05,
    use_colnames=True
)

frequent_itemsets = frequent_itemsets.sort_values(
    by="support",
    ascending=False
)

print("Number of frequent itemsets:", len(frequent_itemsets))

frequent_itemsets.head(15)
```

```python
rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.30
)

rules = rules.sort_values(
    by="lift",
    ascending=False
)

print("Number of association rules:", len(rules))

rules[[
    "antecedents",
    "consequents",
    "support",
    "confidence",
    "lift"
]].head(15)
```

### Task 5 – Recommendation System 

Build a recommendation function using Apriori results.
***Example:***
	recommend("Burger")
**Output**:
&#x9;Fries
Soft Drink
Wrap

```python
def recommend(food_item, top_n=3):
    recommendations = rules[
        rules["antecedents"].apply(
            lambda x: food_item in x
        )
    ].copy()

    recommendations = recommendations.sort_values(
        by="lift",
        ascending=False
    )

    recommended_items = []

    for _, row in recommendations.iterrows():
        for item in row["consequents"]:
            if item != food_item and item not in recommended_items:
                recommended_items.append(item)

        if len(recommended_items) >= top_n:
            break

    return recommended_items[:top_n]
```

### Task 6 – Time Series Forecasting 

- Aggregate monthly orders/sales
- Analyze trends and seasonality
- Build an ARIMA model
- Forecast next 3 months

```python
monthly_data = df.set_index("OrderDate").resample("ME").agg({
    "OrderID": "count",
    "TotalAmount": "sum"
})

monthly_data.columns = ["MonthlyOrders", "MonthlySales"]

print("Monthly orders and sales:")
monthly_data
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))

plt.plot(
    monthly_data.index,
    monthly_data["MonthlySales"],
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    monthly_data.index,
    monthly_data["MonthlyOrders"],
    marker="o"
)

plt.title("Monthly Orders Trend")
plt.xlabel("Month")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

```python
from statsmodels.tsa.arima.model import ARIMA
```

```python
sales_series = monthly_data["MonthlySales"]

model = ARIMA(
    sales_series,
    order=(1, 1, 1)
)

model_fit = model.fit()

print("ARIMA model fitted successfully.")
```

```python
forecast = model_fit.forecast(steps=3)

forecast.index = pd.date_range(
    start=monthly_data.index[-1] + pd.offsets.MonthEnd(1),
    periods=3,
    freq="ME"
)

print("Next 3 Months Sales Forecast:")
print(forecast)
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    sales_series.index,
    sales_series.values,
    marker="o",
    label="Actual Sales"
)

plt.plot(
    forecast.index,
    forecast.values,
    marker="o",
    linestyle="--",
    label="Forecast"
)

plt.title("Actual vs Forecasted Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)
plt.show()
```
